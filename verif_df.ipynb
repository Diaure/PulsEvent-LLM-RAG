```python
import pandas as pd
```

```python
df_verif = pd.read_csv("./data/ge_events_df.csv")
df_verif.head(5)
```

```python
# récupérer les vraies dates de debut et fin de l'évènement
import json
def extract_first_timing(cell):
    if cell is None:
        return None

    # Si c'est une string JSON → on la charge
    if isinstance(cell, str):
        try:
            timings = json.loads(cell)
        except Exception:
            return None
    else:
        # Si c'est déjà une liste Python
        timings = cell

    # si la liste est vide
    if not timings:
        return None, None

    # On prend la première occurrence
    t0 = timings[0]
    return t0.get("begin")

df_verif["begin"] = df_verif["timings"].apply(extract_first_timing)
df_verif["begin"] = pd.to_datetime(df_verif["begin"], utc=True)
```
